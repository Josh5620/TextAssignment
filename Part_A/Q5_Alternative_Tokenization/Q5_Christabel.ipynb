# Q5: Alternative Tokenisation using CountVectorizer (Christabel)

## Setup and Data Loading

```python
from pathlib import Path
from textwrap import fill

import nltk
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import CountVectorizer


nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)


def print_tokens(tokens, width=205):
    # Match the previous token-list output using ordinary print and line wrapping.
    wrapped_tokens = fill(
        repr(tokens),
        width=width,
        subsequent_indent=" ",
        break_long_words=False,
        break_on_hyphens=False,
    )
    print(wrapped_tokens)


def find_data_file(filename):
    # Find Part_A/data from the notebook folder, Part_A, or the repository root.
    marker = Path("data") / filename
    for candidate in [Path.cwd(), *Path.cwd().parents]:
        if (candidate / marker).exists():
            return candidate / marker
        if (candidate / "Part_A" / marker).exists():
            return candidate / "Part_A" / marker
    raise FileNotFoundError(f"Could not locate Part_A/data/{filename}")


DATA_FILE = find_data_file("Data_1.txt")
```

```python
# Read the same text corpus used in Q1 so the outputs can be compared fairly.
with open(DATA_FILE, "r", encoding="utf-8") as file:
    text = file.read()
```

## NLTK Baseline

```python
# Q1 used NLTK word_tokenize, so this is kept as the baseline method.
q1_tokens = word_tokenize(text)
```

## CountVectorizer Alternative

```python
# CountVectorizer is used as the alternative tokenization approach for Q5.
raw_vectorizer = CountVectorizer(
    lowercase=False,
    token_pattern=r"(?u)\b\w+\b",
)
count_vectorizer_tokens = raw_vectorizer.build_analyzer()(text)
```

```python
# Configure CountVectorizer to lowercase the text and remove its English stop words.
# The token pattern extracts word characters, so punctuation is not returned as tokens.
filtered_vectorizer = CountVectorizer(
    stop_words="english",
    token_pattern=r"(?u)\b\w+\b",
)
filtered_count_vectorizer_tokens = filtered_vectorizer.build_analyzer()(text)
```

## Tokenization Results

```python
# Print the original Q1 approach and token count for comparison.
print("=== Q1 Approach: NLTK word_tokenize ===")
print_tokens(q1_tokens)
print(f"Token count: {len(q1_tokens)}")
```

```python
# Print the raw alternative tokenization output.
print("\n=== Q5 Alternative: CountVectorizer Analyzer ===")
print_tokens(count_vectorizer_tokens)
print(f"Token count: {len(count_vectorizer_tokens)}")
```

```python
# Print the cleaned CountVectorizer output after stop-word and punctuation removal.
print("\n=== Q5 Alternative After Stop Word and Punctuation Removal ===")
print_tokens(filtered_count_vectorizer_tokens)
print(f"Filtered token count: {len(filtered_count_vectorizer_tokens)}")
```
