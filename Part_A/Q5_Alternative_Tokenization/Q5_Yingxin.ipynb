# Q5: Alternative Tokenisation using TextBlob (Chung Ying Xin)

This notebook compares the Q1 NLTK baseline with TextBlob as an alternative tokenisation approach.

### Setup and Data Loading

The shared text corpus and required tokenisation tools are loaded for a fair comparison.

```python
import string
from collections import Counter
from pathlib import Path

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from textblob import TextBlob

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("stopwords", quiet=True)

DATA_FILE = Path("../data/Data_1.txt")

with open(DATA_FILE, "r", encoding="utf-8") as f:
    text = f.read()
```

### NLTK Baseline

NLTK `word_tokenize()` reproduces the original Q1 method and provides the baseline tokens.

```python
# Group approach from Q1, kept here as the baseline to compare against.
nltk_tokens = word_tokenize(text)

print("NLTK word_tokenize (group Q1 approach)")
print(nltk_tokens)
print("Token count:", len(nltk_tokens))
```

### TextBlob Alternative

TextBlob's word tokenizer extracts word tokens from the same corpus for comparison with NLTK.

```python
# Alternative approach: TextBlob's .words property.
textblob_tokens = list(TextBlob(text).words)

print("TextBlob (alternative approach)")
print(textblob_tokens)
print("Token count:", len(textblob_tokens))
```

### Tokenization Comparison

Token frequencies are compared to identify differences between the NLTK and TextBlob outputs.

```python
# Counter rather than set, so the four full stops count as four differences and not one.
only_in_nltk = Counter(nltk_tokens) - Counter(textblob_tokens)
only_in_textblob = Counter(textblob_tokens) - Counter(nltk_tokens)

print("Tokens NLTK produces that TextBlob does not:", sum(only_in_nltk.values()))
print(dict(only_in_nltk))

print("\nTokens TextBlob produces that NLTK does not:", sum(only_in_textblob.values()))
print(dict(only_in_textblob))

print("\nHyphenated word check")
print("  'open-class' kept whole by NLTK     :", "open-class" in nltk_tokens)
print("  'open-class' kept whole by TextBlob :", "open-class" in textblob_tokens)

print(f"\nNLTK {len(nltk_tokens)} - {sum(only_in_nltk.values())} punctuation"
      f" = TextBlob {len(textblob_tokens)}")
```

### Is TextBlob a Different Tokeniser?

The analysis checks whether TextBlob produces genuinely different word tokens or mainly excludes punctuation.

```python
# If TextBlob is really NLTK plus a punctuation strip, then stripping punctuation from
# the edges of the NLTK tokens by hand should rebuild the TextBlob list exactly.
stripped = [tok.strip(string.punctuation) for tok in nltk_tokens]
stripped = [tok for tok in stripped if tok]

print("NLTK tokens with edge punctuation stripped:", len(stripped))
print("Reproduces the TextBlob list exactly:", stripped == textblob_tokens)
```

### Filtered Token Comparison

Stop words and punctuation are removed before comparing the meaningful tokens retained by both approaches.

```python
stop_words = set(stopwords.words("english"))
punctuations = set(string.punctuation)

# Same filter as Q1 for the NLTK list. The TextBlob list only needs the stop word step,
# because TextBlob has already dropped the punctuation.
q1_filtered = [
    tok for tok in nltk_tokens
    if tok.lower() not in stop_words and tok not in punctuations
]
q5_filtered = [tok for tok in textblob_tokens if tok.lower() not in stop_words]

print("TextBlob tokens after stop word removal:")
print(q5_filtered)

print("\nQ1 (NLTK, stop words and punctuation removed):", len(q1_filtered), "tokens")
print("Q5 (TextBlob, stop words removed)            :", len(q5_filtered), "tokens")
print("Identical token lists:", q1_filtered == q5_filtered)
```
