# Q5: Alternative Tokenisation using BERT WordPiece (Genemor)

This notebook compares the Q1 NLTK baseline with BERT WordPiece subword tokenisation.

## Setup and Data Loading

The shared text corpus and required tokenisers are loaded for a fair comparison.

```python
from pathlib import Path

from nltk.tokenize import word_tokenize
from transformers import AutoTokenizer


DATA_FILE = Path.cwd().parents[0] / "data" / "Data_1.txt"

with open(DATA_FILE, "r", encoding="utf-8") as f:
    text = f.read()
```

## NLTK Baseline

NLTK `word_tokenize()` reproduces the original Q1 method and provides the baseline tokens.

```python
# NLTK word_tokenize was the Q1 method, kept here as the baseline
nltk_tokens = word_tokenize(text)
```

## BERT WordPiece Alternative

WordPiece divides unfamiliar or complex words into smaller subword units that can be represented by BERT's vocabulary.

```python
# BERT WordPiece as the alternative (subword tokenizer, marks split pieces with ##)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
wordpiece_tokens = tokenizer.tokenize(text)
```

## Tokenization Results

The outputs are compared to highlight words that WordPiece divided into multiple subword tokens.

```python
# pull out the pieces it split mid-word so the subword behaviour is easy to see
subwords = [t for t in wordpiece_tokens if t.startswith("##")]

print("Q1 Approach: NLTK word_tokenize")
print("============================================================")
print(nltk_tokens)
print("Token count:", len(nltk_tokens))

print("\nQ5 Alternative: BERT WordPiece")
print("============================================================")
print(wordpiece_tokens)
print("Token count:", len(wordpiece_tokens))

print("\nSubword pieces (## continuations)")
print("============================================================")
print(subwords)
```
