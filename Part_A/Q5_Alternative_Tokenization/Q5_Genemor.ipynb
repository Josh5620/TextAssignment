# Q5_Genemor

```python
from pathlib import Path

from nltk.tokenize import word_tokenize
from transformers import AutoTokenizer


DATA_FILE = Path.cwd().parents[0] / "data" / "Data_1.txt"

with open(DATA_FILE, "r", encoding="utf-8") as f:
    text = f.read()
```

```python
# NLTK word_tokenize was the Q1 method, kept here as the baseline
nltk_tokens = word_tokenize(text)
```

```python
# BERT WordPiece as the alternative (subword tokenizer, marks split pieces with ##)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
wordpiece_tokens = tokenizer.tokenize(text)
```

```python
# pull out the pieces it split mid-word so the subword behaviour is easy to see
subwords = [t for t in wordpiece_tokens if t.startswith("##")]

print("Q1 Approach: NLTK word_tokenize")
print("============================================================")
print(nltk_tokens)
print("Token count:", len(nltk_tokens))

print("\nQ5 Alternative: BERT WordPiece")
print("============================================================")
print(wordpiece_tokens)
print("Token count:", len(wordpiece_tokens))

print("\nSubword pieces (## continuations)")
print("============================================================")
print(subwords)
```
