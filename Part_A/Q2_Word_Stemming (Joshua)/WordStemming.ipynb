# Part A Q2: Word Stemming

This notebook applies three stemming methods to the same tokens and compares how each method reduces words to their stems.

## Setup and Data Preparation

The text is loaded and the first two sentences are tokenised to provide a consistent input for all three stemmers.

```python
import nltk
from pathlib import Path
from nltk.stem import RegexpStemmer, PorterStemmer, LancasterStemmer
from nltk.tokenize import sent_tokenize, word_tokenize


DATA_FILE = Path.cwd().parents[0] / "data" / "Data_1.txt"
```

```python
# Load text data
with open(DATA_FILE, "r", encoding="utf-8") as f:
    text = f.read()
```

```python
# Extract first 2 sentences and tokenize
sentences = sent_tokenize(text)[:1]
words = word_tokenize(" ".join(sentences))
```

## Regular Expression Stemmer

The regular-expression stemmer removes suffixes that match a predefined pattern.

```python
# Regular Expression Stemmer
reg_stemmer = RegexpStemmer('ing$|ed$|s$|ness$', min=4)
print("=== Regular Expression Stemmer ===")
for word in words:
    print(f"{word} -> {reg_stemmer.stem(word)}")
```

## Porter Stemmer

The Porter stemmer applies a sequence of established rules to produce moderate and commonly used word stems.

```python
# Porter Stemmer
porter = PorterStemmer()
print("\n=== Porter Stemmer ===")
for word in words:
    print(f"{word} -> {porter.stem(word)}")
```

## Lancaster Stemmer

The Lancaster stemmer applies more aggressive rules, which often produce shorter stems than the other methods.

```python
# Lancaster Stemmer
lancaster = LancasterStemmer()
print("\n=== Lancaster Stemmer ===")
for word in words:
    print(f"{word} -> {lancaster.stem(word)}")
```
